## GIT and GITHUB



# commands

1. git init -- Suppose we have a project in local and we want to push it on github, we will use this git init in the terminal to create a brand new empty repo on the github and will do git add . git commit and then push
2. git clone -- when we have a repo on github and we want to have it our system we use git clone (link of the repo).
3. git add -- add changes to staged changes
4. git commit -- changes are ready to push with a message
5. git push -- push the changes to a remote repo
6. git pull -- push the changes from a remote repo
7. git fetch -- get changes in local but does not merge it, git pull = git fetch + git merge
