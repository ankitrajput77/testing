# commands

1. git init -- Suppose we have a project in local and we want to push it on github, we will use this git init in the terminal to create a brand new empty repo on the github and will do git add . git commit and then push
2. git clone -- when we have a repo on github and we want to have it our system we use git clone (link of the repo).
3. git add -- add changes to staged changes
4. git commit -- changes are ready to push with a message
5. git push -- push the changes to a remote repo
6. git pull -- push the changes from a remote repo
7. git fetch -- get changes in local but does not merge it, git pull = git fetch + git merge
8. git branch -- check all the branches
9. git checkout -- go to the particular commit or branch, create new branch, choose their or our changes
10. git switch -- nothing but replace of git checkout
11. git merge -- merge brances into one
12. git reset -- go to the prev commits



## CI/CD - Continious integration/Continious Delivery/Deployment

practices that automate how code moves from a developer's machine to production

#### Continuous Integration (CI)

Developers merge their code changes into a shared repository frequently (often multiple times a day). Each merge automatically triggers:

1. Build — compiling the code to make sure it actually builds
2. Automated tests — running unit/integration tests to catch bugs early
3. Code quality checks — linting, static analysis, etc.

The goal is to catch problems immediately, before they pile up. Instead of everyone working in isolation for weeks and hitting a nightmare of merge conflicts, small changes get integrated and validated constantly.

#### Continuous Delivery vs Continuous Deployment (the two "CD"s)

These are often confused, but they're different:

1. Continuous Delivery — every change that passes CI is automatically prepared for release (packaged, tested further) but a human still clicks a button to actually release it to production.
2. Continuous Deployment — goes one step further and automatically pushes every change that passes all tests straight to production, no human approval needed.
