```python
# polymorphism
from abc import ABC, abstractmethod
class person(ABC):
    def __init__ (self, name, height):
        self.name = name
        self.height = height
    def get_height(self):
        print(self.height)

    @abstractmethod
    def get_realname(self):
        print("this has to be in sub class")


class men(person):
    def get_name(self):
        print(self.name)
    def chromo(self):
        print("XY")
    def get_realname(self):
            print("this has to be in sub class")

class women(person):
    def chromo(self):
        print("XX")
    def get_realname(self):
            print("this has to be in sub class")
```

```python
ob1 = men("Ankit", 5.6)
ob2 = women("Bhumi", 5.4)

ob = [ob1, ob2]

for ob in ob:
    ob.chromo()
```

```python
# class variable
from datetime import datetime
class reals:
    var1 = "okok" ## class variable, cannot be modify outside of classmethod

    @classmethod
    def classfunction(cls):
        print(datetime.fromtimestamp)
        cls.var1 = "kkk"

    @staticmethod
    def staticfunction():
        print("no need of self or anything")
    
```

```python
ob = reals()
print(ob.var1)
ob.classfunction()
print(ob.var1)
ob.staticfunction()
```

```python
## operator overloading

class p:
    def __init__ (self, a, b):
        self.a = a
        self.b = b
    def __add__ (p1, p2):
        return p(p1.a+p2.a, p1.b+p2.b)
    def __eq__(p1, p2):
        return p1.a == p2.a and p1.b == p2.b
    def __repr__(self):
        return f"point is {self.a}"
```

```python
ob = p(1,2)
```

```python
print(ob)
```

```python
# Decorator

## let's decorate it
def decoratorr(func):
    def wrapper(lt):
        lt = func(lt)
        print(lt, "your list has been reversed")
    return wrapper

# I have a function
@decoratorr
def reversee(list:list)->list:
    list.reverse()
    return list

reversee([1,2,3,4])
```

```python
reversee([1,2,3,4])
```
