## Docker

- containerization of the project so that it can be runned anywhere.

### Virtualization vs containerization

- Virtual machines is nothing but a physical computer and inside it some part of it is assigned seperately and it will have its own Operating system, ram, cpu, rom and all.
- Continers are nothing but more light weight environment which will have everything you need to run the code, you just have to ship this container to someone and they will run it without any issue.

every problem solved by containers can be solved by VMs. VMs are overhead solution of the problem where containers is more sensable.

sometimes we need 100 projects to run for that we cannot create 100VMs but we can have 100 containers.

- Docker Image = Docker image is a blueprint of the container, it is kind of template from which the containers are created.
- Docker Contaier =  A container is a running instance of an image. Image = class and Container = Object
- Docker Engine = Now we need something that actually runs and manages containers. That's where Docker Engine comes in. Docker Engine is the core technology that allows you to: create containers, start containers, stop containers, remove containers, build images, manage networks, manage storage, etc. Think of Docker Engine as the main machinery behind Docker.
- Docker Daemon = Docker daemon (dockerd) is a background process that actually listens for Docker commands and performs the work. when we do docker run, CLI passes this command to docker daemon and it works on it in background.
- Docker CLI = It is command line interface, that's it.

- Note that docker engine is not the app, we use docker client (docker desktop) to show all the images, containers but it also go to docker engine and find all the information.

- There are two options to use docker:
1. We can have it in our local.
2. we can use cloud like AWS EC2 serives, which provides virtual machines to install docker and all.

- To run it on local:

- two options:
1. share the code with DockerFile then other person can run the docker build command and can run the application in their system
2. do not share the code with them, push you image on docker hub and the next person can use that image and can run the application in their system.


- why to run it on aws ec2?
- The reason we run it on ec2 is because, to make it simpler and error free.
- to run it, we make instance on EC2, 

- We use compose docker file (docker-compose.yml) to create network between images. 
- we push our images on registry like docker hub so that we can use the images anywhere.
- docker tag will make the pointer pointing to an image. (we tell which image has to be pushed)
- docker push will push the image to hub (we tell in which repo the tagged image will do)
- docker pull will take image from hub
